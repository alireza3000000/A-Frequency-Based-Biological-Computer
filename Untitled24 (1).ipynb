```python
import requests
import json

# تنظیمات اولیه
base_url = "https://chrompolymerdb.bme.uic.edu"
gene = "MYC"
cell_line = "Calu3"

# لیست مسیرهای احتمالی API برای تست
possible_endpoints = [
    f"{base_url}/api/query",
    f"{base_url}/api/data",
    f"{base_url}/api/structure",
    f"{base_url}/query",
    f"{base_url}/data",
    f"{base_url}/api/v1/query",
]

# پارامترهای درخواست
params = {
    "gene": gene,
    "cell_line": cell_line,
    "chromosome": "chr8",
    "start": 127200000,
    "end": 127750000
}

# تست مسیرهای مختلف
for endpoint in possible_endpoints:
    try:
        print(f"در حال تست: {endpoint}")
        response = requests.get(endpoint, params=params, timeout=10)
        
        if response.status_code == 200:
            print(f"✅ موفق! مسیر درست: {endpoint}")
            data = response.json()
            
            # ذخیره داده‌ها
            filename = f"{gene}_{cell_line}_data.json"
            with open(filename, "w") as f:
                json.dump(data, f, indent=2)
            print(f"داده‌ها در {filename} ذخیره شد.")
            break
        else:
            print(f"❌ خطا: {response.status_code}")
            
    except requests.exceptions.RequestException as e:
        print(f"❌ خطای اتصال: {e}")

else:
    print("\n⚠️ هیچ‌کدام از مسیرهای API کار نکردند.")
    print("لطفاً مسیر API را از تب Network مرورگر پیدا کن.")
    print("1. سایت را باز کن: https://chrompolymerdb.bme.uic.edu/")
    print("2. کلید F12 را بزن و به تب Network برو.")
    print("3. یک جستجو انجام بده (مثلاً MYC و Calu3).")
    print("4. درخواستی که با /api/ شروع می‌شود را پیدا کن.")
    print("5. آدرس کامل آن را برای من بفرست.")
```

```python
import os
import requests
from tqdm import tqdm

# تنظیمات
output_dir = "MCF7_HiC_data"
os.makedirs(output_dir, exist_ok=True)

# لینک‌های دانلود داده‌های Single-cell Hi-C (MCF7)
# این لینک‌ها از ENA Project PRJNA802208 استخراج شده‌اند
urls = [
    "https://ftp.sra.ebi.ac.uk/vol1/fastq/SRR170/000/SRR17000000/SRR17000000_1.fastq.gz",
    "https://ftp.sra.ebi.ac.uk/vol1/fastq/SRR170/000/SRR17000000/SRR17000000_2.fastq.gz",
]

# دانلود فایل‌ها
for url in urls:
    filename = url.split("/")[-1]
    filepath = os.path.join(output_dir, filename)
    
    if os.path.exists(filepath):
        print(f"فایل {filename} قبلاً دانلود شده است.")
        continue
    
    print(f"در حال دانلود {filename}...")
    response = requests.get(url, stream=True)
    total_size = int(response.headers.get('content-length', 0))
    
    with open(filepath, 'wb') as f, tqdm(
        desc=filename,
        total=total_size,
        unit='iB',
        unit_scale=True,
        unit_divisor=1024,
    ) as bar:
        for data in response.iter_content(chunk_size=1024):
            size = f.write(data)
            bar.update(size)

print("دانلود کامل شد.")
```

```python
import h5py
import numpy as np
import scipy.sparse as sp
import os

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")
protein_path = os.path.join(paper18, "GSE253721_Blood_RNAADT_Final.h5ad")

with h5py.File(protein_path, 'r') as f:
    # خواندن codes و categories
    codes = f['var']['assay']['codes'][:]
    categories = f['var']['assay']['categories'][:]
    
    # پیدا کردن ایندکس‌های ADT
    adt_idx = np.where(codes == 0)[0]  # ADT = 0
    rna_idx = np.where(codes == 1)[0]  # RNA = 1
    
    print(f"تعداد ADT: {len(adt_idx)}")
    print(f"تعداد RNA: {len(rna_idx)}")
    
    # خواندن ماتریس Sparse
    data = f['X']['data'][:]
    indices = f['X']['indices'][:]
    indptr = f['X']['indptr'][:]
    
    n_cells = len(indptr) - 1
    n_features = len(codes)
    
    X_full = sp.csr_matrix((data, indices, indptr), shape=(n_cells, n_features))
    print(f"\n✅ ماتریس کامل: {X_full.shape}")
    
    # جدا کردن ADT
    X_adt = X_full[:, adt_idx]
    X_rna = X_full[:, rna_idx]
    
    print(f"✅ ADT: {X_adt.shape}")
    print(f"✅ RNA: {X_rna.shape}")
    
    # ذخیره
    sp.save_npz(os.path.join(paper18, "protein_adt.npz"), X_adt)
    sp.save_npz(os.path.join(paper18, "rna_geo.npz"), X_rna)
    print("\n✅ ذخیره شد: protein_adt.npz و rna_geo.npz")
    
    # ذخیره نام پروتئین‌ها
    name_0 = f['var']['name_0'][:]
    adt_names = name_0[adt_idx]
    rna_names = name_0[rna_idx]
    
    np.save(os.path.join(paper18, "protein_names.npy"), adt_names)
    np.save(os.path.join(paper18, "rna_geo_names.npy"), rna_names)
    print("✅ نام‌ها ذخیره شد")
```

```python
import numpy as np
import scipy.sparse as sp
import os

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")

# بارگذاری k-NN
distances = np.load(os.path.join(paper18, "knn_distances.npy"))
indices = np.load(os.path.join(paper18, "knn_indices.npy"))

print(f"distances: {distances.shape}")
print(f"indices: {indices.shape}")

# ===== ساخت شبکه شباهت =====
n_cells = distances.shape[0]
k = distances.shape[1]

# ساخت ماتریس شباهت Sparse
rows = np.repeat(np.arange(n_cells), k)
cols = indices.flatten()
data = 1 - distances.flatten()  # شباهت = ۱ - فاصله

# حذف خود-حلقه‌ها
mask = rows != cols
rows = rows[mask]
cols = cols[mask]
data = data[mask]

# ماتریس شباهت
similarity = sp.csr_matrix((data, (rows, cols)), shape=(n_cells, n_cells))
print(f"\n✅ شبکه شباهت: {similarity.shape}")
print(f"  - تعداد یال‌ها: {similarity.nnz}")

# ذخیره
sp.save_npz(os.path.join(paper18, "similarity_network.npz"), similarity)
print("✅ ذخیره شد: similarity_network.npz")
```

```python
import scipy.sparse as sp
import numpy as np
import os

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")

# بارگذاری شبکه شباهت
similarity = sp.load_npz(os.path.join(paper18, "similarity_network.npz"))
print(f"شبکه: {similarity.shape}, یال‌ها: {similarity.nnz}")

# ===== خوشه‌بندی Louvain =====
import networkx as nx
from networkx.algorithms.community import louvain_communities

# تبدیل به گراف
G = nx.from_scipy_sparse_array(similarity)
print(f"گراف: {G.number_of_nodes()} گره, {G.number_of_edges()} یال")

# Louvain
communities = louvain_communities(G, seed=42)
print(f"\n✅ تعداد خوشه‌ها: {len(communities)}")

# اندازه خوشه‌ها
sizes = [len(c) for c in communities]
print(f"اندازه خوشه‌ها (اولین ۱۰): {sizes[:10]}")
print(f"حداکثر: {max(sizes)}, حداقل: {min(sizes)}")

# ذخیره برچسب خوشه‌ها
labels = np.zeros(G.number_of_nodes(), dtype=int)
for i, community in enumerate(communities):
    for node in community:
        labels[node] = i

np.save(os.path.join(paper18, "cluster_labels.npy"), labels)
print("✅ ذخیره شد: cluster_labels.npy")
```

```python
import numpy as np
import scipy.sparse as sp
import os

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")

# بارگذاری
rna_norm = sp.load_npz(os.path.join(paper18, "rna_norm.npz"))
gene_names = np.load(os.path.join(paper18, "rna_geo_names.npy"), allow_pickle=True)
labels = np.load(os.path.join(paper18, "cluster_labels.npy"))

print(f"RNA: {rna_norm.shape}")
print(f"ژن‌ها: {len(gene_names)}")
print(f"خوشه‌ها: {len(np.unique(labels))}")

# ===== پیدا کردن ژن‌های نشانگر هر خوشه =====
n_clusters = len(np.unique(labels))

for cluster_id in range(min(5, n_clusters)):  # اولین ۵ خوشه
    # سلول‌های این خوشه
    mask = labels == cluster_id
    cluster_cells = rna_norm[mask]
    
    # میانگین بیان
    mean_expr = np.array(cluster_cells.mean(axis=0)).flatten()
    
    # سلول‌های دیگر
    other_cells = rna_norm[~mask]
    mean_other = np.array(other_cells.mean(axis=0)).flatten()
    
    # تفاوت
    diff = mean_expr - mean_other
    
    # ۱۰ ژن برتر
    top_genes = np.argsort(diff)[-10:][::-1]
    
    print(f"\n🧬 خوشه {cluster_id} (تعداد: {mask.sum()}):")
    for g in top_genes:
        print(f"  - {gene_names[g]}: {diff[g]:.3f}")
```

```python
import numpy as np
import scipy.sparse as sp
import os

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")

# بارگذاری
rna_norm = sp.load_npz(os.path.join(paper18, "rna_norm.npz"))
gene_names = np.load(os.path.join(paper18, "rna_geo_names.npy"), allow_pickle=True)

# ===== امتیاز مسیرهای متابولیکی =====
# مسیر گلیکولیز
glycolysis_genes = ['HK2', 'PFKL', 'PKM', 'LDHA', 'SLC2A1', 'PGK1', 'ENO1']
# مسیر OXPHOS
oxphos_genes = ['ND1', 'SDHA', 'COX4I1', 'ATP5A1', 'UQCRC1']

# پیدا کردن ایندکس ژن‌ها
def find_gene_idx(gene_list, gene_names):
    indices = []
    for g in gene_list:
        for i, name in enumerate(gene_names):
            if name == g.encode() if isinstance(name, bytes) else name == g:
                indices.append(i)
                break
    return indices

glycolysis_idx = find_gene_idx(glycolysis_genes, gene_names)
oxphos_idx = find_gene_idx(oxphos_genes, gene_names)

print(f"گلیکولیز: {len(glycolysis_idx)} ژن پیدا شد")
print(f"OXPHOS: {len(oxphos_idx)} ژن پیدا شد")

# محاسبه امتیاز
glycolysis_score = np.array(rna_norm[:, glycolysis_idx].mean(axis=1)).flatten()
oxphos_score = np.array(rna_norm[:, oxphos_idx].mean(axis=1)).flatten()

print(f"\n✅ امتیاز گلیکولیز: {glycolysis_score.shape}")
print(f"✅ امتیاز OXPHOS: {oxphos_score.shape}")

# ذخیره
np.save(os.path.join(paper18, "glycolysis_score.npy"), glycolysis_score)
np.save(os.path.join(paper18, "oxphos_score.npy"), oxphos_score)
print("✅ ذخیره شد: glycolysis_score.npy, oxphos_score.npy")
```

```python
import numpy as np
import scipy.sparse as sp
import os

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")

# بارگذاری
signature = sp.load_npz(os.path.join(paper18, "signature.npz"))  # RNA + پروتئین
glycolysis = np.load(os.path.join(paper18, "glycolysis_score.npy"))
oxphos = np.load(os.path.join(paper18, "oxphos_score.npy"))

print(f"بردار پایه (RNA + پروتئین): {signature.shape}")
print(f"گلیکولیز: {glycolysis.shape}")
print(f"OXPHOS: {oxphos.shape}")

# ===== ترکیب =====
# تبدیل متابولیت به sparse
metabolism = sp.csr_matrix(np.column_stack([glycolysis, oxphos]))
print(f"متابولیت (sparse): {metabolism.shape}")

# ترکیب نهایی
signature_final = sp.hstack([signature, metabolism], format='csr')
print(f"\n✅ بردار امضای نهایی: {signature_final.shape}")

# ذخیره
sp.save_npz(os.path.join(paper18, "signature_final.npz"), signature_final)
print("✅ ذخیره شد: signature_final.npz")

# ===== کاهش بُعد مجدد =====
from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(n_components=50, random_state=42)
signature_final_pca = svd.fit_transform(signature_final)
print(f"\n✅ PCA نهایی: {signature_final_pca.shape}")

np.save(os.path.join(paper18, "signature_final_pca.npy"), signature_final_pca)
print("✅ ذخیره شد: signature_final_pca.npy")
```

```python
import numpy as np
import scipy.sparse as sp
from sklearn.neighbors import NearestNeighbors
from sklearn.decomposition import TruncatedSVD
import networkx as nx
from networkx.algorithms.community import louvain_communities
import os

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")

# بارگذاری بردار نهایی
signature_final = sp.load_npz(os.path.join(paper18, "signature_final.npz"))
print(f"بردار نهایی: {signature_final.shape}")

# ===== PCA =====
svd = TruncatedSVD(n_components=50, random_state=42)
sig_pca = svd.fit_transform(signature_final)
print(f"PCA: {sig_pca.shape}")

# ===== k-NN =====
knn = NearestNeighbors(n_neighbors=15, metric='cosine', n_jobs=-1)
knn.fit(sig_pca)
distances, indices = knn.kneighbors(sig_pca)
print(f"k-NN: {distances.shape}")

# ===== شبکه شباهت =====
n_cells = distances.shape[0]
k = distances.shape[1]
rows = np.repeat(np.arange(n_cells), k)
cols = indices.flatten()
data = 1 - distances.flatten()
mask = rows != cols
rows, cols, data = rows[mask], cols[mask], data[mask]
similarity = sp.csr_matrix((data, (rows, cols)), shape=(n_cells, n_cells))
print(f"شبکه: {similarity.shape}, یال‌ها: {similarity.nnz}")

# ===== خوشه‌بندی =====
G = nx.from_scipy_sparse_array(similarity)
communities = louvain_communities(G, seed=42)
print(f"\n✅ تعداد خوشه‌ها: {len(communities)}")

# برچسب‌ها
labels_new = np.zeros(G.number_of_nodes(), dtype=int)
for i, community in enumerate(communities):
    for node in community:
        labels_new[node] = i

np.save(os.path.join(paper18, "cluster_labels_final.npy"), labels_new)
print("✅ ذخیره شد: cluster_labels_final.npy")

# ===== مقایسه با قبلی =====
from sklearn.metrics import adjusted_rand_score

labels_old = np.load(os.path.join(paper18, "cluster_labels.npy"))
ari = adjusted_rand_score(labels_old, labels_new)
print(f"\n📊 مقایسه خوشه‌بندی:")
print(f"  - ARI (قبلی vs جدید): {ari:.4f}")
```

```python
import numpy as np
import scipy.sparse as sp
import os

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")

# بارگذاری
rna_norm = sp.load_npz(os.path.join(paper18, "rna_norm.npz"))
gene_names = np.load(os.path.join(paper18, "rna_geo_names.npy"), allow_pickle=True)
labels_new = np.load(os.path.join(paper18, "cluster_labels_final.npy"))
labels_old = np.load(os.path.join(paper18, "cluster_labels.npy"))

print(f"خوشه‌های جدید: {len(np.unique(labels_new))}")
print(f"خوشه‌های قبلی: {len(np.unique(labels_old))}")

# ===== بررسی خوشه‌هایی که تغییر کردند =====
from sklearn.metrics import confusion_matrix

# جدول توافق
cm = confusion_matrix(labels_old, labels_new)
print(f"\n✅ ماتریس توافق: {cm.shape}")

# ===== تفسیر ۵ خوشه جدید =====
n_clusters = len(np.unique(labels_new))

for cluster_id in range(min(5, n_clusters)):
    mask = labels_new == cluster_id
    cluster_cells = rna_norm[mask]
    mean_expr = np.array(cluster_cells.mean(axis=0)).flatten()
    other_cells = rna_norm[~mask]
    mean_other = np.array(other_cells.mean(axis=0)).flatten()
    diff = mean_expr - mean_other
    top_genes = np.argsort(diff)[-10:][::-1]
    
    print(f"\n🧬 خوشه {cluster_id} (تعداد: {mask.sum()}):")
    for g in top_genes:
        print(f"  - {gene_names[g]}: {diff[g]:.3f}")
```

```python
import numpy as np
import scipy.sparse as sp
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
import os

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")

# بارگذاری
signature_pca = np.load(os.path.join(paper18, "signature_final_pca.npy"))
labels = np.load(os.path.join(paper18, "cluster_labels_final.npy"))

print(f"PCA: {signature_pca.shape}")
print(f"خوشه‌ها: {len(np.unique(labels))}")

# ===== معیارهای خوشه‌بندی =====
print("\n📊 معیارهای خوشه‌بندی:")

# ۱. Silhouette (بالاتر بهتر)
sil = silhouette_score(signature_pca, labels, sample_size=5000, random_state=42)
print(f"  - Silhouette: {sil:.4f}")

# ۲. Calinski-Harabasz (بالاتر بهتر)
ch = calinski_harabasz_score(signature_pca, labels)
print(f"  - Calinski-Harabasz: {ch:.2f}")

# ۳. Davies-Bouldin (پایین‌تر بهتر)
db = davies_bouldin_score(signature_pca, labels)
print(f"  - Davies-Bouldin: {db:.4f}")

# ذخیره
np.save(os.path.join(paper18, "silhouette.npy"), np.array([sil]))
np.save(os.path.join(paper18, "calinski_harabasz.npy"), np.array([ch]))
np.save(os.path.join(paper18, "davies_bouldin.npy"), np.array([db]))
print("\n✅ ذخیره شد: silhouette.npy, calinski_harabasz.npy, davies_bouldin.npy")
```

```python
import requests
import os
from tqdm import tqdm

paper18 = os.path.join(os.path.expanduser("~"), "Desktop", "paper18")
filepath = os.path.join(paper18, "pbmc_granulocyte_sorted_10k_filtered_feature_bc_matrix.h5")

# حذف فایل خراب
if os.path.exists(filepath):
    os.remove(filepath)
    print("🗑️ فایل خراب حذف شد")

# دانلود مجدد با هدر
url = "https://cf.10xgenomics.com/samples/cell-arc/1.0.0/pbmc_granulocyte_sorted_10k/pbmc_granulocyte_sorted_10k_filtered_feature_bc_matrix.h5"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}

print("⬇️ در حال دانلود مجدد...")
response = requests.get(url, stream=True, headers=headers)
total_size = int(response.headers.get('content-length', 0))

with open(filepath, 'wb') as f, tqdm(
    desc="PBMC-10k",
    total=total_size,
    unit='iB',
    unit_scale=True,
) as bar:
    for data in response.iter_content(chunk_size=1024):
        size = f.write(data)
        bar.update(size)

# بررسی امضا
with open(filepath, 'rb') as f:
    header = f.read(8)
    print(f"\n✅ امضای فایل: {header}")

size = os.path.getsize(filepath) / (1024**2)
print(f"✅ حجم فایل: {size:.2f} MB")
```
